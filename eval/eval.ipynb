# MailPilot eval charts

Reads the CSVs written by `eval/run_eval.py` and plots them. Needs `pip install matplotlib`. Run `run_eval.py` first.

```python
import csv
from pathlib import Path
import matplotlib.pyplot as plt

RESULTS = Path('..') / 'results'

def read(name):
    path = RESULTS / name
    return list(csv.DictReader(open(path))) if path.exists() else []
```

```python
# Retrieval: recall@k per mode
rows = read('retrieval.csv')
ks = ['recall@1', 'recall@3', 'recall@5']
fig, ax = plt.subplots(figsize=(6, 3.5))
for r in rows:
    ax.plot([1, 3, 5], [float(r[k]) for k in ks], marker='o', label=r['mode'])
ax.set_xlabel('k'); ax.set_ylabel('recall@k'); ax.set_ylim(0, 1.05); ax.legend(); ax.set_title('Retrieval')
fig.savefig(RESULTS / 'retrieval.png', dpi=150, bbox_inches='tight')
```

```python
# Classification accuracy and macro-F1 per model
rows = [r for r in read('classification.csv') if r['accuracy'] != 'skipped']
fig, ax = plt.subplots(figsize=(6, 3.5))
x = range(len(rows))
ax.bar([i - 0.2 for i in x], [float(r['accuracy']) for r in rows], width=0.4, label='accuracy')
ax.bar([i + 0.2 for i in x], [float(r['macro_f1']) for r in rows], width=0.4, label='macro-F1')
ax.set_xticks(list(x), [r['model'] for r in rows], rotation=20, ha='right'); ax.set_ylim(0, 1); ax.legend()
fig.savefig(RESULTS / 'classification.png', dpi=150, bbox_inches='tight')
```

```python
# Drafts: RAG vs no-RAG judge scores (only after an LLM run)
rows = [r for r in read('drafts.csv') if r['faithfulness'] != 'skipped']
for r in rows:
    print(r['model'], r['condition'], 'faithfulness', r['faithfulness'], 'helpfulness', r['helpfulness'])
```
